```python
import json
```

```python
# Load The Data
def load_data(filename):
    with open(filename, 'r') as f:
        data = json.load(f)
    return data
```

```python
data =load_data('store_data.json')
print(data)
print(type(data))
```

```python
#  Clean and S tructure the data
def clean_data(data):
    text_to_num = {"one":1, "two":2,"three":3,"four":4,"five":5}
    cleaned_data = []
    unique_user = set()

    for user in data:
        # Cleaning ratting part
        row_rating = str(user["rating"]).strip().lower()
        if row_rating in text_to_num:
            row_rating = text_to_num[row_rating]
        # print(row_rating)
        user["rating"] = row_rating
        
        # Handale missing value
        row_age = user.get("age")
        if row_age == None:
            user["age"] = None
        # Duplicate data

        if user["name"].strip() in unique_user:
            continue
        cleaned_data.append(user)
    return cleaned_data
```

```python
clean_data(data)
```
