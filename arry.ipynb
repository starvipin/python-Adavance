```python
import array as arr
```

```python

val = arr.array('i',[1,2,3,4,5,6,])

for i in range(0,len(val)):
    print(val[i], end = ' ')


print(type(val))
print(val.typecode)
```

```python
# reverse the array

val.reverse()

for i in range(0,len(val)):
    print(val[i], end = ' ')
```

```python
# inserting a new value in the array

val.insert(1, 50)

print(val)
```

```python
# add last eliment in the array
val.append(100)
print(val)
```

```python

val = arr.array('i',[1,2,3,4,5,6,])
copyarray = arr.array(val.typecode, [x * 2 for x in val])

for i in range(0,len(val)): 
    print(copyarray[i], end = ' ')
```

```python
#  delete arrry eliments
val = arr.array('i',[1,2,3,4,5,6,])
copyarray = arr.array(val.typecode, [x for x in val])

copyarray.pop(3)

for i in range(0,len(copyarray)): 
    print(copyarray[i], end = ' ')
```

```python
#  reverse the array
val = arr.array('i',[1,2,3,4,5,6,])
copyarray = arr.array(val.typecode, [x for x in val])

a = copyarray[::-1]

for i in range(0,len(a)): 
    print(a[i], end = ' ')
```

```python
# user input in array

val = arr.array('i',[])

elements = int(input("Enter How meny elements insert:= "))

for i in range(0,elements): 
    val.append(int(input("Enter the elements:= ")))             

for x in val: 
    print(x, end = ' ')
    
```
