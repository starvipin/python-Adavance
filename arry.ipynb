```python
import array as arr

val = arr.array('i',[1,2,3,4,5,6,])

for i in range(0,len(val)):
    print(val[i], end = ' ')


print(type(val))
print(val.typecode)
```

```python
# reverse the array

val.reverse()

for i in range(0,len(val)):
    print(val[i], end = ' ')
```

```python
# inserting a new value in the array

val.insert(1, 50)
```
