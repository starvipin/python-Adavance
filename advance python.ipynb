# *args
is used to pass a variable number of positional arguments to a function. It collects all the positional arguments into a tuple.

```python
def add(*numbers):
    totle  = 0
    for i in numbers:
        totle += i
    return totle

print(add(1, 2, 3, 4, 5))  # Output: 15
```

# **keyword
is used to pass a variable number of keyword arguments to a function. It collects all the keyword arguments into a dictionary

```python
def create_user(**user_info):
    return user_info

print(create_user(name="Alice", age=30, city="New York"))  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
```
