# *args
is used to pass a variable number of positional arguments to a function. It collects all the positional arguments into a tuple.

```python
def add(*numbers):
    totle  = 0
    for i in numbers:
        totle += i
    return totle

print(add(1, 2, 3, 4, 5))  # Output: 15
```

# **keyword
is used to pass a variable number of keyword arguments to a function. It collects all the keyword arguments into a dictionary

```python
def create_user(**user_info):
    return user_info

print(create_user(name="Alice", age=30, city="New York"))  # Output: {'name': 'Alice', 'age': 30, 'city': 'New York'}
```

# Try exception and finally block

```python
# Try exception and finally block

try:
    x = int(input("Enter number:= "))
    result = 10/x

except ZeroDivisionError:
    print("you connot divide by zero")
except ValueError:
    print("Invalid input. Please enter a valid integer.")

else:
    print("Result is:", result)
    
finally:
    print("Execution completed.")
```

#  custom exception handling

```python
#  custom exception handling

class invalidAgeError(Exception):
    pass

def check_age(age):
    if age < 18:
        raise invalidAgeError("Age cannot be less than 18.")
    print("Valid age:", age)

# check_age(5)  # This will raise the custom exception

try:
    check_age(10)
except invalidAgeError as e:
    print("custom error Caught", e)    

```

```python
# ex

class InsufficientBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance for withdrawal.")
        self.balance -= amount
        print(f"Withdrawal successful. New balance: {self.balance}")

try:
    acc = BankAccount(2000)
    acc.withdraw(1000)  # Successful withdrawal
except InsufficientBalanceError as e:
    print("Custom error caught:", e)
```
