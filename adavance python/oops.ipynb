OOPs kya hota hai?

"OOPs stands for Object-Oriented Programming. It is a programming approach where we organize code using objects and classes. It helps us make code more reusable, maintainable, and easier to manage."

OOPs ke 4 main pillars hote hain:

1. Encapsulation – Encapsulation means wrapping data and methods together inside a class and controlling access to the data.(private virible (__) )
2. Inheritance   – Inheritance allows one class to acquire the properties and methods of another class. It helps in code reusability.
3. Polymorphism  – Polymorphism means the same method or interface can have different behaviors depending on the object or situation.
4. Abstraction   – Abstraction means hiding unnecessary implementation details and showing only the essential functionality to the user.

```python
class Student:

    school = "VIT school"
    # Blueprint for creating a student object
    def __init__(self, name , marks, attendance):
        self.name = name
        self.marks = marks
        self.attendance = attendance

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 80:
            return "B"
        elif self.marks >= 70:
            return "C"
        else:
            return "D"
```

```python
s1 = Student("Alice", 85, 90)
s2 = Student("Bob", 78, 80)
print(Student.school)
print(s1.name)
print(s2.name)
print(s1.calculate_grade())
print(s2.calculate_grade())
```

# encapsulation
Encapsulation means wrapping data and methods together inside a class and controlling access to the data. using private veriable (__)

```python
class BankAccount:

    Bank = "VIT Bank"

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance")

    def get_balance(self):
        return self.__balance
```

```python
acc1 = BankAccount("vipin", 10000)
print('Initial balance: ', acc1.get_balance())
acc1.deposit(5000)
print('After deposit 5000 : ', acc1.get_balance())
acc1.withdraw(2000)
print('After withdrawal 2000: ', acc1.get_balance())

# acc1.balance = 2000000
# print(acc1.name, acc1.balance)
# print(acc1.__balance)
```

# Inheritance   
Inheritance allows one class to acquire the properties and methods of another class. It helps in code reusability.


```python
class Animal:

    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")


class Dog(Animal):

    def bark(self):
        print(f"{self.name} is barking")


dog = Dog("Tommy")

dog.eat()      # Inherited from Animal
dog.sleep()    # Inherited from Animal
dog.bark()     # Dog's own method
```

# Polymorphism  
Polymorphism means the same method or interface can have different behaviors depending on the object or situation.

```python
class Animal:

    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):

    def speak(self):
        print("Dog says: Woof!")

class Cat(Animal):

    def speak(self):
        print("Cat says: Meow!")

class Cow(Animal):

    def speak(self):
        print("Cow says: Moo!")

dog = Dog()
cat = Cat()
cow = Cow()

dog.speak()
cat.speak()
cow.speak()
```

# Abstraction
Abstraction means hiding unnecessary implementation details and showing only the essential functionality to the user.

```python
from abc import ABC, abstractmethod


# Abstract Parent Class
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass

    def sleep(self):
        print("Animal is sleeping")

# Child Class
class Dog(Animal):

    def make_sound(self):
        print("Dog says: Woof!")

# Child Class
class Cat(Animal):

    def make_sound(self):
        print("Cat says: Meow!")

# Objects
dog = Dog()
cat = Cat()

dog.make_sound()
dog.sleep()

cat.make_sound()
cat.sleep()
```
