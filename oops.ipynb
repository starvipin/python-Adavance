OOPs kya hota hai?

"OOPs stands for Object-Oriented Programming. It is a programming approach where we organize code using objects and classes. It helps us make code more reusable, maintainable, and easier to manage."

OOPs ke 4 main pillars hote hain:

1. Encapsulation – Data aur methods ko ek single unit (class) mein bind karna.
2. Inheritance – Ek class doosri class ke properties aur methods ko inherit kar sakti hai.
3. Polymorphism – Same method/interface ka different situations mein different behavior.
4. Abstraction – Unnecessary implementation details ko hide karke sirf required functionality dikhana.

```python
class Student:

    school = "VIT school"
    # Blueprint for creating a student object
    def __init__(self, name , marks, attendance):
        self.name = name
        self.marks = marks
        self.attendance = attendance

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 80:
            return "B"
        elif self.marks >= 70:
            return "C"
        else:
            return "D"
```

```python
s1 = Student("Alice", 85, 90)
s2 = Student("Bob", 78, 80)
print(Student.school)
print(s1.name)
print(s2.name)
print(s1.calculate_grade())
print(s2.calculate_grade())
```
